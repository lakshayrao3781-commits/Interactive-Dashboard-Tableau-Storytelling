Q1 Explain the difference between a Tableau Dashboard and a Tableau Story Point. Give two healthcare-related
use cases.

Tableau Dashboard:Definition: A single, interactive canvas that combines multiple worksheets (charts) to give an overview of live metrics.   Use Case 1 (Live Bed & ICU Monitoring): Real-time tracking of operational bed occupancy, emergency room patient queue, and staff allocation across wards.

Tableau Story Point:Definition: A sequence of interactive visualization views arranged step-by-step in a narrative flow to guide stakeholders through analysis.   Use Case 2 (Annual Hospital Performance Review): Sequential presentation showing patient load trends, treatment cost breakdown, regional patient outcomes, and strategic action recommendations for management.

Q2 What are Actions in Tableau? Explain how Filter Action can help hospital management analyze patient data.

Actions in Tableau: Interactive features (Filter, Highlight, URL, Parameter, Set) that enable interactive user responses between worksheets and dashboards.   

Filter Action Application:Clicking on a specific region (e.g., West) automatically filters all associated charts (e.g., Department Costs, Outcome status) to display data for that selected region.Allows hospital leadership to instantly isolate localized operational issues without configuring multiple individual filters.
